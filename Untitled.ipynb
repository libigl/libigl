```python
import itertools

people = ['Chenxi', 'Yuta', 'Lily', 'Lulu', 'Selena']
groupings = set()

for group2 in itertools.combinations(people, 2):
    group3 = tuple(sorted(set(people) - set(group2)))
    group2 = tuple(sorted(group2))
    # Store as a frozenset of frozensets to avoid duplicates
    grouping = frozenset([frozenset(group2), frozenset(group3)])
    groupings.add(grouping)

# Print results
for g in groupings:
    g2, g3 = sorted(g, key=len)
    print(f"{sorted(g2)}\n{sorted(g3)}")
```
